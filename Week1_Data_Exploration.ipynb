## 1. Problem Statement

Government agencies collect large amounts of digital-service and grievance data. This notebook demonstrates a reproducible workflow to explore workload, disposal performance and resolution delays.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option('display.max_columns', None)
df = pd.read_csv('../data/raw/illustrative_e_governance_data.csv')
df.head()
```

## 2. Data Inspection

```python
print('Shape:', df.shape)
print('\nData types:')
print(df.dtypes)
print('\nMissing values:')
print(df.isna().sum())
print('\nDuplicate rows:', df.duplicated().sum())
```

## 3. Cleaning and Derived Metric

```python
numeric_cols = ['Grievances_Received', 'Grievances_Disposed', 'Avg_Resolution_Days', 'Digital_Service_Score']
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df = df.drop_duplicates()
df['Disposal_Rate_Percent'] = (df['Grievances_Disposed'] / df['Grievances_Received'] * 100).round(2)
df
```

## 4. Validation Checks

```python
checks = {
    'missing_values': int(df.isna().sum().sum()),
    'negative_received': int((df['Grievances_Received'] < 0).sum()),
    'negative_disposed': int((df['Grievances_Disposed'] < 0).sum()),
    'disposed_greater_than_received': int((df['Grievances_Disposed'] > df['Grievances_Received']).sum()),
    'invalid_disposal_rate': int(((df['Disposal_Rate_Percent'] < 0) | (df['Disposal_Rate_Percent'] > 100)).sum())
}
checks
```

## 5. Descriptive Statistics

```python
df[['Grievances_Received','Grievances_Disposed','Disposal_Rate_Percent','Avg_Resolution_Days','Digital_Service_Score']].describe().round(2)
```

## 6. Received vs Disposed

```python
x = np.arange(len(df))
plt.figure(figsize=(10,5))
plt.bar(x - 0.2, df['Grievances_Received'], width=0.4, label='Received')
plt.bar(x + 0.2, df['Grievances_Disposed'], width=0.4, label='Disposed')
plt.xticks(x, df['State_UT'], rotation=45, ha='right')
plt.ylabel('Number of grievances')
plt.title('Illustrative Grievance Volume: Received vs Disposed')
plt.legend()
plt.tight_layout()
plt.show()
```

## 7. Disposal Rate

```python
tmp = df.sort_values('Disposal_Rate_Percent', ascending=False)
plt.figure(figsize=(10,5))
plt.bar(tmp['State_UT'], tmp['Disposal_Rate_Percent'])
plt.xticks(rotation=45, ha='right')
plt.ylabel('Disposal rate (%)')
plt.title('Illustrative Disposal Rate by State/UT')
plt.ylim(0, 105)
plt.tight_layout()
plt.show()
```

## 8. Resolution Time

```python
tmp = df.sort_values('Avg_Resolution_Days')
plt.figure(figsize=(10,5))
plt.bar(tmp['State_UT'], tmp['Avg_Resolution_Days'])
plt.xticks(rotation=45, ha='right')
plt.ylabel('Average resolution time (days)')
plt.title('Illustrative Average Resolution Time')
plt.tight_layout()
plt.show()
```

## 9. Exploratory Correlation

```python
corr = df[['Disposal_Rate_Percent','Avg_Resolution_Days','Digital_Service_Score']].corr().round(2)
corr
```

### Interpretation

Correlation is used here only as an exploratory tool. With a small illustrative dataset, it cannot establish causation. A verified dataset with a meaningful time period and consistent definitions should be used for statistical conclusions.

## 10. Initial Findings to Test Later

- High workload may be associated with longer resolution times.
- Disposal rate should be interpreted together with resolution time.
- Digital-service capability may be associated with better service outcomes, but this requires proper testing.
- Data definitions and reporting practices can strongly influence comparisons.
